```python
# Student marks dataset

students = {
    "Rahul": 92,
    "Ananya": 85,
    "Arjun": 76,
    "Priya": 68,
    "Kiran": 55,
    "Sneha": 43
}

print("Student Marks:")
for name, marks in students.items():
    print(name, ":", marks)
```

```python
# Function to calculate grade

def calculate_grade(marks):
    if marks < 0 or marks > 100:
        return "Invalid Marks"
    elif marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 50:
        return "D"
    else:
        return "F"


# Calculate and display grades

print("Student Grade Report:")
print("-" * 35)

for name, marks in students.items():
    grade = calculate_grade(marks)
    print(f"{name}: Marks = {marks}, Grade = {grade}")
```

```python
# Testing the grade calculator with different marks

test_marks = [95, 82, 74, 65, 58, 45, 30, 101, -5]

print("Testing Grade Calculator:")
print("-" * 35)

for marks in test_marks:
    grade = calculate_grade(marks)
    print(f"Marks = {marks} → Grade = {grade}")
```

## Results

The Student Grade Calculator successfully assigns grades based on student marks.

### Grading Criteria

| Below 0 or Above 100 | Invalid Marks |
