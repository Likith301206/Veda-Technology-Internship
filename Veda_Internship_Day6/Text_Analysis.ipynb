### Objective

The objective of this task is to analyze a given text using Python and calculate the number of words, characters, vowels, and repeated words.

### Tools Used

- Python
- Jupyter Notebook

### Approach

The program uses Python string operations to process the given text. The `split()` function is used to separate words, `len()` is used to count characters and words, and a dictionary is used to identify repeated words. A loop is used to count vowels.

```python
# Text dataset

text = "Python is easy. Python is powerful. Python is useful for data analysis. Python is fun."

print("Text:")
print(text)
```

```python
# Count words and characters

words = text.split()

word_count = len(words)
character_count = len(text)

print("Number of Words:", word_count)
print("Number of Characters:", character_count)
```

```python
# Count vowels

vowels = "aeiouAEIOU"
vowel_count = 0

for character in text:
    if character in vowels:
        vowel_count += 1

print("Number of Vowels:", vowel_count)
```

```python
# Find repeated words

word_frequency = {}

for word in words:
    word = word.lower().strip(".,!?")
    word_frequency[word] = word_frequency.get(word, 0) + 1

repeated_words = {}

for word, count in word_frequency.items():
    if count > 1:
        repeated_words[word] = count

print("Repeated Words:", repeated_words)
```

# Results

The given text was successfully analyzed using Python.

- **Number of Words:** 16
- **Number of Characters:** 81
- **Number of Vowels:** 23
- **Repeated Words:** Python (4 times), is (3 times)
