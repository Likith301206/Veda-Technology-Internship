```python
students = [
    {"id": 101, "name": "Rahul", "marks": 92, "age": 20, "department": "CSE"},
    {"id": 102, "name": "Ananya", "marks": 85, "age": 19, "department": "ISE"},
    {"id": 103, "name": "Arjun", "marks": 76, "age": 21, "department": "CSE"},
    {"id": 104, "name": "Priya", "marks": 68, "age": 20, "department": "ECE"},
    {"id": 105, "name": "Kiran", "marks": 55, "age": 22, "department": "ME"},
    {"id": 106, "name": "Sneha", "marks": 88, "age": 19, "department": "CSE"},
    {"id": 107, "name": "Vikram", "marks": 73, "age": 21, "department": "ISE"},
    {"id": 108, "name": "Neha", "marks": 95, "age": 20, "department": "CSE"}
]

print("Student Records:")
for student in students:
    print(student)
```

```python
high_scorers = []

for student in students:
    if student["marks"] >= 80:
        high_scorers.append(student)

print("Students with marks 80 or above:")

for student in high_scorers:
    print(student)
```

```python
cse_students = []

for student in students:
    if student["department"] == "CSE":
        cse_students.append(student)

print("CSE Students:")

for student in cse_students:
    print(student)
```

```python
young_students = []

for student in students:
    if student["age"] <= 20:
        young_students.append(student)

print("Students aged 20 or below:")

for student in young_students:
    print(student)
```

```python
filtered_students = []

for student in students:
    if student["marks"] >= 80 and student["department"] == "CSE":
        filtered_students.append(student)

print("CSE students with marks 80 or above:")

for student in filtered_students:
    print(student)
```

```python
print("Filtered Student Names:")

for student in filtered_students:
    print(student["name"])
```
