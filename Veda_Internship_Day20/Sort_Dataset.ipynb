```python
import pandas as pd

# Load the Iris dataset
df = pd.read_csv("../Veda_Internship_Day16/iris.csv")

print("Original Dataset:")
display(df.head())

# 1. Sort by sepal length in ascending order
print("\n1. Sepal Length - Ascending Order:")
ascending = df.sort_values(by="sepal_length", ascending=True)
display(ascending.head())

# 2. Sort by sepal length in descending order
print("\n2. Sepal Length - Descending Order:")
descending = df.sort_values(by="sepal_length", ascending=False)
display(descending.head())

# 3. Sort by petal length in ascending order
print("\n3. Petal Length - Ascending Order:")
display(df.sort_values(by="petal_length", ascending=True).head())

# 4. Sort by multiple columns
print("\n4. Sort by Species and Sepal Length:")
multiple_sort = df.sort_values(
    by=["species", "sepal_length"],
    ascending=[True, False]
)
display(multiple_sort.head(10))

print("\nDataset sorting completed successfully!")
```
