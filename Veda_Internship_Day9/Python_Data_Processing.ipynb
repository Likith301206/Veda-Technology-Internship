```python
records = [
    {"id": 101, "name": "Rahul", "department": "CSE", "salary": 35000},
    {"id": 102, "name": "Ananya", "department": "ISE", "salary": 42000},
    {"id": 103, "name": "Arjun", "department": "CSE", "salary": 30000},
    {"id": 104, "name": "Priya", "department": "ECE", "salary": 45000},
    {"id": 105, "name": "Kiran", "department": "ISE", "salary": 38000},
    {"id": 106, "name": "Sneha", "department": "CSE", "salary": 50000},
    {"id": 107, "name": "Vikram", "department": "ECE", "salary": 32000},
    {"id": 108, "name": "Neha", "department": "CSE", "salary": 48000}
]

print("Total Records:", len(records))
```

```python
print("Employee Records")

for record in records:
    print(record)
```

```python
def calculate_average_salary(records):
    total_salary = sum(record["salary"] for record in records)
    return total_salary / len(records)

average_salary = calculate_average_salary(records)

print("Average Salary:", average_salary)
```

```python
salaries = [record["salary"] for record in records]

print("Highest Salary:", max(salaries))
print("Lowest Salary:", min(salaries))
```

```python
def filter_by_department(records, department):
    return [
        record for record in records
        if record["department"] == department
    ]

cse_employees = filter_by_department(records, "CSE")

print("CSE Employees:")

for employee in cse_employees:
    print(employee)
```

```python
def generate_summary(records):
    salaries = [record["salary"] for record in records]

    summary = {
        "total_employees": len(records),
        "average_salary": sum(salaries) / len(salaries),
        "highest_salary": max(salaries),
        "lowest_salary": min(salaries)
    }

    return summary

summary = generate_summary(records)

print("Employee Summary Statistics:")

for key, value in summary.items():
    print(key.replace("_", " ").title() + ":", value)
```

```python
sorted_records = sorted(records, key=lambda record: record["salary"], reverse=True)

print("Employees Sorted by Salary (Highest to Lowest):")

for record in sorted_records:
    print(record["name"], "-", record["salary"])
```

```python
print("Processed Employee Records")

for record in records:
    print(
        "ID:", record["id"],
        "| Name:", record["name"],
        "| Department:", record["department"],
        "| Salary:", record["salary"]
    )
```
