```python
import pandas as pd
import numpy as np

# Load the Iris dataset
df = pd.read_csv("../Veda_Internship_Day16/iris.csv")

# Introduce missing values for demonstration
df.loc[0, "sepal_length"] = np.nan
df.loc[5, "petal_width"] = np.nan
df.loc[10, "species"] = np.nan

# Display the dataset
print("Dataset with Missing Values:")
display(df.head(12))

# Count missing values in each column
print("\nMissing Values in Each Column:")
print(df.isnull().sum())

# Calculate total missing values
total_missing = df.isnull().sum().sum()
print("\nTotal Missing Values:", total_missing)

# Calculate percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("\nMissing Value Percentage by Column:")
print(missing_percentage.round(2))

# Create a missing-value summary
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": missing_percentage.round(2)
})

print("\nMissing Value Summary:")
display(missing_summary)

# Identify columns requiring treatment
columns_with_missing = df.columns[df.isnull().any()].tolist()

print("\nColumns Requiring Treatment:")
print(columns_with_missing)
```
