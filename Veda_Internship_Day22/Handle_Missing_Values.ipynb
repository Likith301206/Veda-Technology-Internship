```python
import pandas as pd
import numpy as np

# Load the Iris dataset
df = pd.read_csv("../Veda_Internship_Day16/iris.csv")

# Introduce missing values for demonstration
df.loc[0, "sepal_length"] = np.nan
df.loc[5, "petal_width"] = np.nan
df.loc[10, "species"] = np.nan

# Display original missing-value counts
print("Original Missing Values:")
print(df.isnull().sum())

# Create a copy for cleaning
cleaned_df = df.copy()

# Fill missing numerical values with their column medians
numerical_columns = cleaned_df.select_dtypes(include="number").columns

for column in numerical_columns:
    cleaned_df[column] = cleaned_df[column].fillna(
        cleaned_df[column].median()
    )

# Fill missing categorical values with the most frequent value
categorical_columns = cleaned_df.select_dtypes(
    include=["object", "str", "category"]
).columns

for column in categorical_columns:
    if cleaned_df[column].isnull().any():
        cleaned_df[column] = cleaned_df[column].fillna(
            cleaned_df[column].mode()[0]
        )

# Display cleaned dataset
print("\nCleaned Dataset:")
display(cleaned_df.head(12))

# Compare missing values before and after cleaning
print("\nMissing Values After Cleaning:")
print(cleaned_df.isnull().sum())

print("\nTotal Missing Values Before Cleaning:",
      df.isnull().sum().sum())

print("Total Missing Values After Cleaning:",
      cleaned_df.isnull().sum().sum())

# Explain the selected treatment
print("\nTreatment Explanation:")
print("Numerical columns: Missing values were replaced with column medians.")
print("Categorical columns: Missing values were replaced with the most frequent category.")
print("The median is less sensitive to extreme values than the mean.")
print("The mode preserves a valid category instead of removing the record.")
```
