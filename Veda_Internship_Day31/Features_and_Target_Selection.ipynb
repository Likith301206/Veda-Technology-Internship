```python
import pandas as pd
from IPython.display import display
```

```python
data = {
    "Hours_Studied": [2, 3, 4, 5, 6, 7, 8, 9, 1, 10],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 95, 50, 98],
    "Previous_Score": [35, 40, 45, 50, 55, 65, 70, 75, 30, 85],
    "Final_Marks": [38, 43, 48, 55, 62, 70, 78, 85, 32, 92]
}

df = pd.DataFrame(data)

print("Student Performance Dataset:")
display(df)
```

```python
X = df[["Hours_Studied", "Attendance", "Previous_Score"]]

print("Selected Features (X):")
display(X)
```

```python
y = df["Final_Marks"]

print("Selected Target (y):")
display(y)
```

```python
print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nFeature Columns:")
print(X.columns.tolist())

print("\nTarget Column:")
print(y.name)

print("\nMissing Values in Features:")
print(X.isnull().sum())

print("\nMissing Values in Target:")
print(y.isnull().sum())
```

```python
print("FEATURE AND TARGET SELECTION")
print("--------------------------------")

print("Features (X):")
print("1. Hours_Studied")
print("2. Attendance")
print("3. Previous_Score")

print("\nTarget (y):")
print("Final_Marks")

print("\nReason for Selection:")
print("1. Study hours may help explain student performance.")
print("2. Attendance may provide useful information about participation.")
print("3. Previous scores may help estimate future performance.")
print("4. Final marks are the value the model will attempt to predict.")
print("5. The target column is kept separate from the input features.")
```
