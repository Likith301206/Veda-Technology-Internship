```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Create a sample dataset containing an outlier
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Marks": [45, 50, 55, 60, 62, 65, 68, 70, 72, 150]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

# Calculate quartiles and IQR
Q1 = df["Marks"].quantile(0.25)
Q3 = df["Marks"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# Identify outliers
outliers = df[
    (df["Marks"] < lower_bound) |
    (df["Marks"] > upper_bound)
]

print("\nPotential Outliers:")
display(outliers)

# Create box plot
plt.figure(figsize=(8, 5))
plt.boxplot(df["Marks"])
plt.title("Box Plot of Student Marks")
plt.ylabel("Marks")
plt.grid(True)
plt.show()

# Observations
print("\nObservations:")
print("1. A mark of 150 is identified as a potential outlier.")
print("2. Most marks fall between 45 and 72.")
print("3. Outliers should be investigated before deciding whether to remove them.")
```
