### Objective

The objective of this task is to analyze a list of numbers using Python and identify the even numbers, odd numbers, largest number, and smallest number.

### Tools Used

- Python
- Jupyter Notebook

### Approach

The program uses a Python list to store numbers. A `for` loop and the modulo operator `%` are used to separate even and odd numbers. The built-in `max()` and `min()` functions are used to find the largest and smallest values.

```python
# Number dataset

numbers = [12, 7, 25, 18, 3, 42, 9, 16, 31, 5]

print("Numbers:")
print(numbers)
```

```python
# Find even and odd numbers

even_numbers = []
odd_numbers = []

for number in numbers:
    if number % 2 == 0:
        even_numbers.append(number)
    else:
        odd_numbers.append(number)

print("Even Numbers:", even_numbers)
print("Odd Numbers:", odd_numbers)
```

```python
# Find largest and smallest numbers

largest_number = max(numbers)
smallest_number = min(numbers)

print("Largest Number:", largest_number)
print("Smallest Number:", smallest_number)
```

# Results

The number dataset was successfully analyzed using Python.

- **Total Numbers:** 10
- **Even Numbers:** 12, 18, 42, 16
- **Odd Numbers:** 7, 25, 3, 9, 31, 5
- **Largest Number:** 42
- **Smallest Number:** 3

# Conclusion

The program successfully identifies even and odd numbers and finds the largest and smallest numbers from a given list.

This task demonstrates the use of Python lists, `for` loops, `if-else` conditions, the modulo operator `%`, and built-in functions such as `max()` and `min()`.
