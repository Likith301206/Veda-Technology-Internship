```python
import pandas as pd

# Load the Iris dataset
df = pd.read_csv("../Veda_Internship_Day16/iris.csv")

print("Original Dataset:")
display(df.head())

# Example 1: Sepal length greater than 5
condition1 = df[df["sepal_length"] > 5]
print("\n1. Sepal length greater than 5:")
display(condition1.head())
print("Number of records:", len(condition1))

# Example 2: Petal length greater than 4
condition2 = df[df["petal_length"] > 4]
print("\n2. Petal length greater than 4:")
display(condition2.head())
print("Number of records:", len(condition2))

# Example 3: Apply two conditions using AND
condition3 = df[
    (df["sepal_length"] > 5) &
    (df["petal_length"] > 4)
]
print("\n3. Sepal length > 5 AND petal length > 4:")
display(condition3.head())
print("Number of records:", len(condition3))

# Example 4: Filter setosa flowers
condition4 = df[df["species"] == "setosa"]
print("\n4. Setosa flowers:")
display(condition4.head())
print("Number of records:", len(condition4))
```
