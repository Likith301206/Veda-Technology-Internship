# Employee Salary Analysis

## Veda Technology Internship — Day 3

### Objective

The objective of this task is to analyze employee salary data using Python and calculate the total, average, highest, and lowest salary.

### Tools Used

- Python
- Jupyter Notebook

### Approach

The employee salary values are stored in a Python list. Built-in Python functions such as `sum()`, `max()`, and `min()` are used to perform the required calculations.

```python
# Employee salary dataset

salaries = [25000, 32000, 28000, 45000, 38000, 52000, 30000, 41000, 35000, 60000]

print("Employee Salaries:")
print(salaries)
```

```python
# Calculate salary statistics

total_salary = sum(salaries)
average_salary = total_salary / len(salaries)
highest_salary = max(salaries)
lowest_salary = min(salaries)

print("Total Salary:", total_salary)
print("Average Salary:", average_salary)
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)
```

# Results

The employee salary data was successfully analyzed using Python.

- **Number of Employees:** 10
- **Total Salary:** ₹3,96,000
- **Average Salary:** ₹39,600
- **Highest Salary:** ₹60,000
- **Lowest Salary:** ₹25,000

# Conclusion

The program successfully calculates the total, average, highest, and lowest salaries from a list of employee salary data using Python.
