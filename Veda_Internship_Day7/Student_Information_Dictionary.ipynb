```python
students = {
    101: {"name": "Rahul", "age": 20, "course": "CSE"},
    102: {"name": "Ananya", "age": 19, "course": "ISE"},
    103: {"name": "Arjun", "age": 21, "course": "CSE"},
    104: {"name": "Priya", "age": 20, "course": "ECE"},
    105: {"name": "Kiran", "age": 19, "course": "AIML"}
}

print("Student Information:")
for student_id, details in students.items():
    print(student_id, details)
```

```python
student_id = 103

if student_id in students:
    print("Student Found:")
    print(students[student_id])
else:
    print("Student not found.")
```

```python
new_student_id = 106

students[new_student_id] = {
    "name": "Sneha",
    "age": 20,
    "course": "ISE"
}

print("New student added successfully!")
print(students[new_student_id])
```

```python
student_id = 103

students[student_id]["age"] = 22
students[student_id]["course"] = "AI & ML"

print("Student information updated successfully!")
print(students[student_id])
```

```python
print("Updated Student Information:")

for student_id, details in students.items():
    print("ID:", student_id)
    print("Name:", details["name"])
    print("Age:", details["age"])
    print("Course:", details["course"])
    print("--------------------")
```
