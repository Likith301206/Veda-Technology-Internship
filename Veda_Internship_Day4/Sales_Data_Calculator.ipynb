### Objective

The objective of this task is to analyze a collection of sales values using Python and calculate the total sales, average sales, highest sale, and lowest sale.

### Tools Used

- Python
- Jupyter Notebook

### Approach

The sales values are stored in a Python list. Built-in Python functions such as `sum()`, `max()`, and `min()` are used to perform the required calculations. The average sales value is calculated by dividing the total sales by the number of sales values.

```python
# Sales data

sales = [15000, 22000, 18000, 25000, 30000, 12000, 28000, 20000, 35000, 17000]

print("Sales Values:")
print(sales)
```

```python
# Calculate sales statistics

total_sales = sum(sales)
average_sales = total_sales / len(sales)
highest_sale = max(sales)
lowest_sale = min(sales)

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```

# Results

The sales data was successfully analyzed using Python.

- **Number of Sales:** 10
- **Total Sales:** ₹2,27,000
- **Average Sales:** ₹22,700
- **Highest Sale:** ₹35,000
- **Lowest Sale:** ₹12,000

# Conclusion

The program successfully calculates the total, average, highest, and lowest sales values from a list of sales data using Python.
