```python
import numpy as np

# Create a numerical dataset
data = np.array([45, 60, 72, 85, 90, 55, 78, 95, 68, 82])

# Display the dataset
print("Dataset:", data)

# Calculate statistical values
print("\nMean:", np.mean(data))
print("Median:", np.median(data))
print("Standard Deviation:", np.std(data))
print("Minimum Value:", np.min(data))
print("Maximum Value:", np.max(data))

# Compare mean and median
mean = np.mean(data)
median = np.median(data)

print("\nMean:", mean)
print("Median:", median)

if mean > median:
    print("The mean is greater than the median.")
elif mean < median:
    print("The mean is less than the median.")
else:
    print("The mean and median are equal.")

# Identify unusually high or low values
print("\nValues greater than 85:", data[data > 85])
print("Values less than 50:", data[data < 50])
```
