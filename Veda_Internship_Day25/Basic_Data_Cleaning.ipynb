```python
import pandas as pd
from IPython.display import display

# Create a sample employee dataset
data = {
    "Employee_ID": ["101", "102", "103", "104", "102", "105", "106", "107"],
    "Name": ["Rahul", "Priya", "Arjun", "Sneha", "Priya", "Kiran", None, "Divya"],
    "Age": ["22", "25", "24", "23", "25", None, "26", "21"],
    "Salary": ["25000", "32000", "28000", "30000", "32000", "27000", "35000", None],
    "Department": ["CSE", "ISE", "CSE", "ECE", "ISE", "AIML", "CSE", "ISE"]
}

df = pd.DataFrame(data)

print("ORIGINAL DATASET")
display(df)

print("\nOriginal Dataset Information:")
df.info()

# Step 1: Check missing values
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())

# Step 2: Remove duplicate records
df = df.drop_duplicates().reset_index(drop=True)

# Step 3: Correct data types
df["Employee_ID"] = pd.to_numeric(df["Employee_ID"], errors="coerce")
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")

# Step 4: Handle missing values
df["Name"] = df["Name"].fillna("Unknown")
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
df["Department"] = df["Department"].fillna(df["Department"].mode()[0])

# Step 5: Display cleaned dataset
print("\nCLEANED DATASET")
display(df)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nCorrected Data Types:")
print(df.dtypes)

print("\nRows in Cleaned Dataset:", len(df))
print("Duplicate Records Remaining:", df.duplicated().sum())
print("Total Missing Values Remaining:", df.isnull().sum().sum())

print("\nBasic Data Cleaning Completed Successfully!")
```
