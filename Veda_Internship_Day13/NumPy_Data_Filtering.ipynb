```python
import numpy as np

# Create a NumPy array
data = np.array([10, 25, 35, 50, 60, 75, 85, 100])

print("Original Array:", data)

# Filter values greater than 50
print("\nValues greater than 50:", data[data > 50])

# Filter values less than 50
print("Values less than 50:", data[data < 50])

# Filter values equal to 50
print("Values equal to 50:", data[data == 50])

# Filter values greater than or equal to 60
print("Values greater than or equal to 60:", data[data >= 60])

# Filter values less than or equal to 35
print("Values less than or equal to 35:", data[data <= 35])
```
