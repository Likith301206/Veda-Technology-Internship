```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Hours_Studied": [2, 3, 4, 5, 6, 7, 8, 4, 6, 9],
    "Attendance": [60, 65, 70, 75, 80, 85, 90, 70, 80, 95],
    "Marks": [40, 45, 50, 60, 65, 75, 85, 55, 70, 95]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

print("\nDataset Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())
```

```python
print("Dataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
df_cleaned = df.drop_duplicates().copy()

df_cleaned = df_cleaned.dropna()

print("Cleaned Dataset:")
display(df_cleaned)

print("\nMissing Values After Cleaning:")
print(df_cleaned.isnull().sum())

print("\nCleaned Dataset Shape:", df_cleaned.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df_cleaned["Marks"], bins=5, edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df_cleaned["Hours_Studied"],
    df_cleaned["Marks"]
)

plt.title("Study Hours vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df_cleaned["Attendance"],
    df_cleaned["Marks"]
)

plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    df_cleaned["Student"],
    df_cleaned["Marks"]
)

plt.title("Marks by Student")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.grid(axis="y")

plt.show()
```

```python
correlation = df_cleaned[
    ["Hours_Studied", "Attendance", "Marks"]
].corr()

print("Correlation Matrix:")
display(correlation)
```

```python
print("OBSERVATIONS")

print("1. The dataset contains information about 10 students.")

print(
    "2. Average marks:",
    round(df_cleaned["Marks"].mean(), 2)
)

print(
    "3. Highest marks:",
    df_cleaned.loc[df_cleaned["Marks"].idxmax(), "Marks"]
)

print(
    "4. Student with highest marks:",
    df_cleaned.loc[df_cleaned["Marks"].idxmax(), "Student"]
)

print(
    "5. Correlation between study hours and marks:",
    round(
        df_cleaned["Hours_Studied"].corr(df_cleaned["Marks"]),
        2
    )
)

print(
    "6. Correlation between attendance and marks:",
    round(
        df_cleaned["Attendance"].corr(df_cleaned["Marks"]),
        2
    )
)
```
