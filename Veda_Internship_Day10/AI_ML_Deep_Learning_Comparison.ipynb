# 1. Artificial Intelligence (AI)

Artificial Intelligence is a broad field of computer science that focuses on creating machines and systems capable of performing tasks that normally require human intelligence.

AI systems can perform tasks such as reasoning, problem-solving, decision-making, and understanding language.

**Examples:**
- Chatbots and virtual assistants
- Chess-playing programs
- Self-driving cars
- Expert systems

**Real-world application:** A virtual assistant that understands questions and provides relevant answers.

# 2. Machine Learning (ML)

Machine Learning is a subset of Artificial Intelligence that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every situation.

ML algorithms improve their performance by learning from examples and data.

**Examples:**
- Email spam detection
- Movie recommendation systems
- Fraud detection
- House price prediction

**Real-world application:** A movie streaming platform that recommends movies based on a user's viewing history.

# 3. Deep Learning (DL)

Deep Learning is a subset of Machine Learning that uses artificial neural networks with multiple layers to learn complex patterns from large amounts of data.

It is especially useful for tasks involving images, audio, video, and natural language.

**Examples:**
- Facial recognition
- Speech recognition
- Image classification
- Advanced language models

**Real-world application:** A facial recognition system that identifies a person from an image.

# 4. Comparison of AI, ML, and Deep Learning

| Feature | Artificial Intelligence (AI) | Machine Learning (ML) | Deep Learning (DL) |
|---|---|---|---|
| Definition | Enables machines to perform tasks requiring human intelligence | Enables machines to learn patterns from data | Uses multi-layer neural networks to learn complex patterns |
| Scope | Broadest field | Subset of AI | Subset of ML |
| Data Requirement | Depends on the approach | Usually requires training data | Often requires large amounts of data |
| Human Intervention | Depends on the system | May require feature engineering | Can learn features automatically |
| Computing Power | Varies by application | Moderate to high | Often high |
| Techniques | Rule-based systems, search, planning | Decision trees, regression, clustering | Deep neural networks, CNNs, Transformers |
| Applications | Planning, expert systems, robotics | Spam detection, recommendations | Image recognition, speech recognition |

# 5. Relationship Between AI, ML, and Deep Learning

Artificial Intelligence is the broad field of creating systems that can perform tasks requiring human intelligence.

Machine Learning is a subset of AI that allows systems to learn patterns from data rather than relying only on explicitly written rules.

Deep Learning is a subset of Machine Learning that uses neural networks with multiple layers to learn complex patterns.

### Relationship

- AI is the broadest field.
- ML is a subset of AI.
- Deep Learning is a subset of ML.

Therefore:

**Artificial Intelligence ⊃ Machine Learning ⊃ Deep Learning**

### Example

Consider a system that recognizes objects in photographs.

- **AI:** The overall goal is to build an intelligent object-recognition system.
- **ML:** The system learns patterns from labeled images.
- **Deep Learning:** A deep neural network learns complex image features and identifies objects.
